```python

import seaborn as sns
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
import pandas as pd
```

```python
df.isnull().sum()
```

```python
df["payment"].mode()[0]
```

```python
df["payment"]=df["payment"].fillna("credit card")
```

```python
df["payment"].isnull().sum()
```

```python
df["pickup_zone"].mode()[0]
```

```python
df["pickup_zone"]=df["pickup_zone"].fillna("Midtown Center")
```

```python
df["dropoff_zone"].mode()
```

```python
df["dropoff_zone"]=df["dropoff_zone"].fillna("Upper East Side North")
```

```python
df["pickup_borough"].mode()[0]
```

```python
df["pickup_borough"]=df["pickup_borough"].fillna("Manhattan")
```

```python
df["dropoff_borough"].mode()[0]
```

```python
df["dropoff_borough"]=df["dropoff_borough"].fillna("Manhattan")
```

```python
df.isnull().sum()
```

```python
df.head(3)
```

```python
df["pickup"]=pd.to_datetime(df["pickup"],errors="coerce")
```
