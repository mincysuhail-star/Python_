❖	Final Insights 

After performing data cleaning, statsistical analysis, and EDA  on socail media engagement analysis dataset containing 5000 entries.Australia has highest engagement rate , its followed by Brazil ,Canada, France.Country USA has least engagement rate.


Post type video has highest average engagement rate and followed by text type.
Image type showing lowest post type.
The relationship between age and engagement rate is very weak.
Users aged 18–24(adult category) show a relatively high median engagement rate.
The 55–64 (aged) age group has the lowest median.
Verified accounts have a slightly higher average engagement rate (0.776398 float) than unverified accounts shows less engagement rate.And it doesn't provide any consistent engagement rate.
Posted_at timing not showing the best hour of the day for impressions from this.Thus need to be provide better offers in weekend or peak timings , such as sunday noon , friday night.
Mobile device shows highest mean engagement rate and following desktop devices and tablet device showing less average rate .Need to be provide more special offers for the device type and for the posting time.
Negative sentiment posts perform best in terms of average engagement and watch time, while neutral posts also maintain relatively strong performance. Positive posts have the lowest average engagement.
